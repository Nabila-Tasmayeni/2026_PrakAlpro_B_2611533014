## 2.1 Dasar Teori
## DASAR TEORI

Python merupakan bahasa pemrograman yang memiliki penulisan kode cukup sederhana sehingga cocok digunakan untuk mempelajari dasar-dasar pemrograman. Dalam pembuatan program, terdapat beberapa struktur yang dapat membantu menjalankan perintah sesuai kebutuhan. Salah satunya adalah perulangan yang memungkinkan suatu perintah dijalankan beberapa kali tanpa harus menuliskan kode yang sama secara berulang.

### Perulangan For

Perulangan for digunakan ketika sebuah proses perlu dilakukan secara berulang berdasarkan suatu urutan atau jumlah tertentu. Pada Python, for sering digunakan bersama range() untuk menentukan berapa kali proses akan dijalankan. Perulangan ini juga dapat digunakan untuk membaca atau memproses data satu per satu.

### Range()

range() digunakan untuk menghasilkan urutan bilangan yang dapat dimanfaatkan dalam perulangan. Penggunaannya dapat disesuaikan dengan kebutuhan, baik menentukan nilai awal, batas akhir, maupun langkah perulangannya.

Bentuk umum range() adalah:

python  
range(batas)  
range(awal, batas)  
range(awal, batas, langkah)  


Nilai pada batas akhir tidak termasuk dalam hasil range(). Jika langkah yang digunakan bernilai negatif, urutan perulangan dapat berjalan dari nilai yang lebih besar menuju nilai yang lebih kecil.

### Fungsi

Dalam sebuah program, terdapat kemungkinan suatu perintah perlu digunakan lebih dari satu kali. Untuk menghindari penulisan kode yang sama secara berulang, Python menyediakan fungsi. Fungsi merupakan bagian program yang dibuat untuk mengerjakan tugas tertentu dan dapat dipanggil kembali saat diperlukan.

Fungsi dibuat menggunakan def dan dapat memiliki parameter untuk menerima data. Jika diperlukan, fungsi juga dapat mengembalikan hasil menggunakan return. Dengan adanya fungsi, bagian-bagian program dapat dipisahkan berdasarkan tugasnya sehingga lebih mudah dipahami.

### Nested For

Ketika satu perulangan membutuhkan perulangan lain di dalam prosesnya, digunakan nested for. Artinya, terdapat for di dalam for lainnya. Pada struktur ini, perulangan bagian dalam akan dijalankan setiap kali perulangan bagian luar melakukan satu proses.

Nested for sering digunakan untuk proses yang memiliki lebih dari satu tingkat, misalnya mengatur baris dan kolom atau membuat susunan pola menggunakan karakter maupun angka.

### If, Elif, dan Else

Selain perulangan, program juga membutuhkan percabangan untuk menentukan perintah berdasarkan kondisi. Pada Python, if digunakan untuk memeriksa suatu kondisi. Jika kondisi tersebut tidak terpenuhi, pemeriksaan dapat dilanjutkan menggunakan elif. Sementara itu, else digunakan ketika tidak ada kondisi sebelumnya yang terpenuhi.

Penggunaan if, elif, dan else dapat dipadukan dengan perulangan. Dengan demikian, setiap proses dalam perulangan dapat menghasilkan tindakan yang berbeda sesuai dengan kondisi yang sedang diperiksa.

### Variabel

Variabel digunakan untuk menyimpan nilai yang dibutuhkan selama program berjalan. Dalam materi perulangan, variabel dapat digunakan untuk menyimpan batas perulangan, nilai penghitung, maupun data yang akan diproses.

Nilai pada variabel dapat berubah selama program dijalankan. Oleh karena itu, variabel membantu program dalam mengatur dan mengolah data sesuai dengan proses yang sedang dilakukan.

### Input

Program dapat dibuat agar menerima data secara langsung dari pengguna dengan menggunakan input(). Masukan tersebut dapat digunakan untuk menentukan nilai yang akan diproses, seperti jumlah perulangan atau nilai tertentu yang akan diperiksa oleh percabangan.

Data hasil input() memiliki tipe string. Jika masukan berupa angka dan akan digunakan dalam perhitungan atau perulangan, data tersebut dapat diubah terlebih dahulu menggunakan int() atau tipe data lain yang sesuai.

### Penggunaan Perulangan dalam Program

Perulangan for dapat dikombinasikan dengan range(), fungsi, dan percabangan untuk membuat program yang memiliki proses lebih teratur. Misalnya, sebuah fungsi dapat digunakan untuk menjalankan suatu proses, kemudian for digunakan untuk mengulanginya beberapa kali. Percabangan if, elif, dan else dapat digunakan untuk menentukan hasil yang sesuai dengan kondisi.

Dengan mempelajari materi tersebut, penggunaan perulangan dalam Python tidak hanya terbatas pada menjalankan perintah berulang, tetapi juga dapat digunakan untuk membuat program yang lebih terstruktur dan memiliki proses yang sesuai dengan kebutuhan.


## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2611533014"

````python
#Buat file dengan nama perulangan_for1_2611533014.py
#Buat program untuk perulangan for dalam python
#Nama variabel ditambah 4 digit NIM terakhir contoh:ulang_1234
#Program ini menggunakan fungsi input ()

ulang_3014 = int(input("Masukkan jumlah perulangan: "))

for i in range (ulang_3014):
    print(f"Perulangan ke-{i}")
````

**Penjelasan :**
1. ulang_3014 = int(input("Masukkan jumlah perulangan: "))  
  Perintah ini digunakan untuk meminta pengguna memasukkan jumlah perulangan. Fungsi input() digunakan untuk menerima masukan dari pengguna, sedangkan int() digunakan untuk mengubah masukan tersebut menjadi bilangan bulat. Nilai yang dimasukkan kemudian disimpan ke dalam variabel ulang_3014.

2. for i in range(ulang_3014):  
  Perintah ini digunakan untuk membuat perulangan for. Variabel i berfungsi sebagai penghitung, sedangkan range(ulang_3014) menentukan jumlah perulangan sesuai dengan angka yang dimasukkan pengguna. Perulangan dimulai dari angka 0.

3. print(f"Perulangan ke-{i}")  
  Perintah ini digunakan untuk menampilkan hasil dari setiap perulangan. Nilai i akan berubah pada setiap proses perulangan sehingga nomor yang ditampilkan juga ikut berubah.

Secara keseluruhan, program ini meminta pengguna menentukan jumlah perulangan, kemudian menjalankan perulangan sebanyak jumlah yang dimasukkan. Jika pengguna memasukkan angka **5**, maka perulangan akan dilakukan sebanyak 5 kali, dengan nilai i dari **0 sampai 4**.

* print(i, end=" ")  
  Perintah ini menampilkan setiap nilai i dalam satu baris dengan jarak satu spasi antarangka.

program ini menunjukkan dua cara penggunaan range(). Perulangan pertama menghasilkan angka dari **0 sampai ulang_3014 - 1**, sedangkan perulangan kedua menghasilkan angka dari **1 sampai ulang_3014**.


### 2.2.3 Program "perulangan_for3_2611533014"

````python
#Buat file dengan nama perulangan_for3_2611533014.py
#Buat program untuk perulangan for dalam python
#Nama variabel ditambah 4 digit NIM terakhir contoh:ulang_1234
#Program ini menggunakan fungsi input ()

ulang_3014 = int(input("Masukkan jumlah perulangan: "))

jumlah_3014 = 0
for i in range(1,ulang_3014+1):
    print(i, end=" ")
    jumlah_3014 = jumlah_3014 + i

    if i < ulang_3014:
        print("+", end="")
    else:
        print("=", jumlah_3014,end="")
print()
print("jumlah =", jumlah_3014)
````

  Variabel jumlah_3014 digunakan untuk menyimpan hasil penjumlahan. Nilainya dimulai dari 0 karena belum ada angka yang dijumlahkan.

3. for i in range(1, ulang_3014+1):  
  Perintah ini membuat perulangan yang dimulai dari angka 1 sampai nilai ulang_3014. Penambahan +1 dilakukan karena batas akhir pada range() tidak ikut dihitung.

4. print(i, end=" ")  
  Perintah ini menampilkan angka yang sedang diproses. end=" " membuat angka tetap berada pada satu baris dan diberi jarak.

5. jumlah_3014 = jumlah_3014 + i  
  Perintah ini digunakan untuk menjumlahkan nilai i dengan nilai yang sudah tersimpan di jumlah_3014. Hasil penjumlahan kemudian disimpan kembali ke dalam variabel tersebut.

6. if i < ulang_3014:  
  Kondisi ini digunakan untuk memeriksa apakah angka yang sedang diproses masih belum mencapai angka terakhir. Jika masih ada angka berikutnya, program akan menampilkan tanda +.

7. print("+", end="")  
  Perintah ini menampilkan tanda + tanpa berpindah ke baris baru. Tanda tersebut digunakan sebagai pemisah antarangka dalam proses penjumlahan.

8. else:  
  Bagian ini dijalankan ketika nilai i sudah mencapai angka terakhir dalam perulangan.

9. print("=", jumlah_3014,end="")  
  Perintah ini menampilkan tanda = dan hasil penjumlahan yang sudah diperoleh.

10. print()  
  Perintah ini digunakan untuk membuat baris baru setelah seluruh proses perulangan selesai.

11. print("jumlah =", jumlah_3014)  
  Perintah ini menampilkan kembali jumlah akhir dari seluruh angka yang telah dijumlahkan.

program ini digunakan untuk menjumlahkan angka dari **1 sampai jumlah perulangan yang dimasukkan**. Misalnya pengguna memasukkan 4, program akan menampilkan proses 1 + 2 + 3 + 4 = 10, kemudian menampilkan **jumlah = 10**.

### 2.2.5 Program "nested_for1_2611533014"

````python
#Buat file dengan nama nested_for1_2611533014.py
#Buat program untuk perulangan for dalam python
#Nama variabel ditambah 4 digit NIM terakhir contoh:ulang_1234
#Program ini menggunakan fungsi input ()

batas_3014 = int(input("Masukkan nilai batas: "))
for line_3014 in range(1,batas_3014 +1):
    for j_3014 in range(1, (-1* line_3014  + batas_3014 + 1)):
        print("*", end=" ")
    print(line_3014)
````

**Penjelasan:**

1. batas_3014 = int(input("Masukkan nilai batas: "))  
  Perintah ini digunakan untuk meminta pengguna memasukkan nilai batas. Nilai yang dimasukkan diubah menjadi bilangan bulat menggunakan int() dan disimpan dalam variabel batas_3014.

2. for line_3014 in range(1, batas_3014 + 1):  
  Perulangan ini digunakan sebagai perulangan bagian luar. Variabel line_3014 berfungsi untuk menentukan baris yang sedang diproses. Perulangan dimulai dari 1 sampai nilai batas_3014.

3. for j_3014 in range(1, (-1 * line_3014 + batas_3014 + 1)):  
  Perulangan ini merupakan perulangan bagian dalam atau nested for. Perulangan ini digunakan untuk menentukan jumlah tanda * yang akan ditampilkan pada setiap baris. Jumlah tanda * akan berkurang satu pada setiap baris berikutnya.

4. print("*", end=" ")  
  Perintah ini digunakan untuk menampilkan tanda *. end=" " membuat tanda * tetap berada pada baris yang sama dan diberi jarak.

5. print(line_3014)  
  Setelah perulangan bagian dalam selesai, perintah ini menampilkan nomor baris. Setelah itu, program berpindah ke baris berikutnya.

program ini menggunakan **nested for** untuk membuat pola tanda * yang jumlahnya semakin berkurang pada setiap baris. Misalnya jika nilai batas yang dimasukkan adalah 5, hasilnya akan seperti:
```text
* * * * 1
* * * 2
* * 3
* 4
5
```

Jadi, semakin besar nomor barisnya, **semakin sedikit tanda * yang ditampilkan**.


### 2.2.6 Program "nested_for2_2611533014"

````python
#Buat file dengan nama nested_for2_2611533014.py
#Buat program untuk perulangan for dalam python
#Nama variabel ditambah 4 digit NIM terakhir contoh:ulang_1234
#Program ini menggunakan fungsi input ()

batas_3014 = int(input("Masukkan nilai batas: "))
for i in range(1, batas_3014 + 1):
    for j in range(1, batas_3014 + 1):
        print("*", end="")
    print() #Pindah ke baris berikutnya
````

**Penjelasan:**


1. batas_3014 = int(input("Masukkan nilai batas: "))  
  Perintah ini digunakan untuk meminta pengguna memasukkan nilai batas. Nilai tersebut diubah menjadi bilangan bulat dengan int() dan disimpan dalam variabel batas_3014.

2. for i in range(1, batas_3014 + 1):  
  Perulangan ini merupakan perulangan bagian luar. Variabel i digunakan untuk menentukan jumlah baris yang akan dibuat. Perulangan dimulai dari 1 sampai nilai batas_3014.

3. for j in range(1, batas_3014 + 1):  
  Perulangan ini merupakan perulangan bagian dalam (nested for). Variabel j digunakan untuk menentukan jumlah tanda * yang dicetak pada setiap baris. Jumlahnya mengikuti nilai batas_3014.

4. print("*", end="")  
  Perintah ini digunakan untuk mencetak tanda *. Penggunaan end="" membuat tanda * tetap berada pada baris yang sama tanpa langsung berpindah ke baris berikutnya.

5. print()  
  Perintah ini digunakan setelah perulangan bagian dalam selesai. Fungsinya untuk berpindah ke baris berikutnya sehingga tanda * dari proses berikutnya dicetak pada baris baru.

program ini menggunakan nested for untuk membuat pola persegi yang terdiri dari tanda *. Perulangan i menentukan jumlah baris, sedangkan perulangan j menentukan jumlah tanda * pada setiap baris.

Misalnya jika memasukkan nilai **4**, hasilnya:

```text
****
****
****
****
```

Jadi, jika nilai batas adalah 4, program membuat 4 baris dan setiap baris berisi 4 tanda *.


### 2.2.7 Program "nested_for3_2611533014"

````python
#Buat file dengan nama nested_for3_2611533014.py
#Buat program untuk perulangan for dalam python
#Nama variabel ditambah 4 digit NIM terakhir contoh:ulang_1234
#Program ini menggunakan fungsi input ()

batas_3014 = int(input("Masukkan nilai batas: "))
for i in range(batas_3014+1):
    for j in range(batas_3014+1):
        print(i+j, end=" ")
    print() #Pindah ke baris berikutnya
````

  Perintah ini digunakan setelah perulangan bagian dalam selesai untuk berpindah ke baris berikutnya. Dengan begitu, hasil dari setiap nilai i akan tersusun dalam baris yang berbeda.

program ini menggunakan nested for untuk membuat susunan angka berdasarkan hasil penjumlahan nilai i dan j. Perulangan i mengatur baris, sedangkan perulangan j mengatur kolom.

Misalnya jika memasukkan nilai 3, hasilnya:

```text

3 4 5 6
```

Jadi, setiap angka yang ditampilkan merupakan hasil dari **i + j** pada posisi baris dan kolom tersebut.


### 2.2.8 Program "nested_for4_2611533014"

````python
#Buat file dengan nama nested_for4_2611533014.py
#Buat program untuk perulangan for dalam python
#Nama variabel ditambah 4 digit NIM terakhir contoh:ulang_1234
#Program ini menggunakan fungsi input ()

tinggi_3014 = int(input("Masukkan tinggi pola (bilangan genap,misal 10): "))

if tinggi_3014 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3014 = tinggi_3014
    c_3014 = a_3014
    lebar_3014 = (2 * tinggi_3014) - 2

    for i_3014 in range(1, tinggi_3014 + 1):
       b_3014 = c_3014 + 1

       for j_3014 in range(1, lebar_3014 + 1):

          #Baris atas dan bawah
          if i_3014 == 1 or i_3014 == tinggi_3014:
             if j_3014 == 1 or j_3014 == lebar_3014:
                print("#", end="")
             else:
                print(" ", end="")

          #Baris isi
          else:
            if j_3014 == 1 or j_3014 == lebar_3014:
                    print("|", end="")
            else:
                    if j_3014 == c_3014:
                        print("<", end="")
                    elif j_3014 == b_3014:
                        print(">", end="")
                    elif j_3014 == (lebar_3014 - c_3014):
                        print("<", end="")
                    elif j_3014 == (lebar_3014 - c_3014 + 1):
                        print(">", end="")
                    elif j_3014 > b_3014 and j_3014 < (lebar_3014 - c_3014):
                        print("=", end="")
                    else:
                        print(" ", end="")
                        
       print()
       
       #logika asli java
       a_3014 -= 2

       if a_3014 <= 0:
           c_3014 = (-a_3014) + 2
       else:
           c_3014 = a_3014
````

Mahasiswa diminta merancang program konsol interaktif untuk mencetak pola bangun geometri dua dimensi berupa Jam Pasir Kristal Palindromik yang terbungkus rapi di dalam bingkai pembatas berpagar. Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.

**Kode Jawaban :**

````python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK ===")

n_3014 = int(input("Masukkan ukuran skala jam pasir (N): "))

if n_3014 <= 0:
    print("N harus bilangan positif!")
else:

    # Border atas
    print("#", end="")
    for i_3014 in range(4 * n_3014 + 5):
        print("=", end="")
    print("#")

    # Jam pasir bagian atas
    for baris_3014 in range(n_3014, 0, -1):

        print("|", end="")

        # Spasi kiri
        print(" ", end="")
        for spasi_3014 in range(2 * (n_3014 - baris_3014)):
            print(" ", end="")

        # Angka menurun
        for angka_3014 in range(baris_3014, 0, -1):
            print(angka_3014, end=" ")

        # Poros kristal
        print("<*>", end="")

        # Angka menaik
        for angka_3014 in range(1, baris_3014 + 1):
            print(" ", end="")
            print(angka_3014, end="")

        # Spasi kanan
        for spasi_3014 in range(2 * (n_3014 - baris_3014)):
            print(" ", end="")

        print(" |")

    # Titik pusat
    print("|", end="")

    for spasi_3014 in range(2 * n_3014 + 1):
        print(" ", end="")

    print("<*>", end="")

    for spasi_3014 in range(2 * n_3014 + 1):
        print(" ", end="")

    print("|")

    # Jam pasir bagian bawah
    for baris_3014 in range(1, n_3014 + 1):

        print("|", end="")

        # Spasi kiri
        print(" ", end="")
        for spasi_3014 in range(2 * (n_3014 - baris_3014)):
            print(" ", end="")

        # Angka menurun
        for angka_3014 in range(baris_3014, 0, -1):
            print(angka_3014, end=" ")

        # Poros kristal
        print("<*>", end="")

        # Angka menaik
        for angka_3014 in range(1, baris_3014 + 1):
            print(" ", end="")
            print(angka_3014, end="")

        # Spasi kanan
        for spasi_3014 in range(2 * (n_3014 - baris_3014)):
            print(" ", end="")

        print(" |")

    # Border bawah
    print("#", end="")
    for i_3014 in range(4 * n_3014 + 5):
        print("=", end="")
    print("#")
````
