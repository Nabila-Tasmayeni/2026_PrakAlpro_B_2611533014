## 2.1 Dasar Teori
Dalam pemrograman, tipe data dan variabel merupakan bagian dasar yang penting karena digunakan untuk menyimpan dan mengolah berbagai macam informasi. Tipe data menentukan jenis nilai yang dapat disimpan, sedangkan variabel digunakan sebagai tempat untuk menyimpan nilai tersebut. Dengan memahami tipe data dan variabel, kita bisa membuat program yang dapat menerima input, melakukan proses, dan menghasilkan output sesuai kebutuhan.

1) String

String adalah tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String biasanya ditulis menggunakan tanda kutip, misalnya \"Nabila\", \"Halo Dunia\", atau \"123\". Walaupun \"123\" terlihat seperti angka, jika ditulis sebagai string maka nilainya dianggap sebagai teks dan bukan angka yang bisa langsung digunakan untuk operasi matematika.

2) Numerik

Tipe data numerik digunakan untuk menyimpan nilai berupa angka. Data numerik biasanya digunakan dalam perhitungan matematika, seperti penjumlahan, pengurangan, perkalian, dan pembagian. Contohnya adalah 10, 25, 3.14, dan 100. Data numerik dapat berupa bilangan bulat maupun bilangan desimal, tergantung kebutuhan program.

3) Input

Input adalah data yang dimasukkan oleh pengguna ke dalam program. Input membuat program menjadi lebih interaktif karena nilai yang digunakan tidak selalu ditentukan dari awal oleh pembuat program. Contohnya, program dapat meminta pengguna memasukkan nama, umur, atau nilai ujian. Data yang diberikan melalui input kemudian dapat disimpan dalam variabel untuk diproses lebih lanjut.
    
4) Konstanta

Konstanta adalah nilai yang tetap dan tidak berubah selama program berjalan. Konstanta biasanya digunakan untuk menyimpan nilai yang memang sudah ditentukan dan tidak perlu diubah, seperti nilai PI = 3.14 atau jumlah hari dalam satu minggu yaitu 7. Penggunaan konstanta dapat membuat program lebih mudah dipahami dan mengurangi kemungkinan terjadinya kesalahan saat program dijalankan.

5) Boolean

Boolean adalah tipe data yang hanya memiliki dua kemungkinan nilai, yaitu True (benar) dan False (salah). Boolean sering digunakan dalam kondisi atau percabangan untuk menentukan apakah suatu pernyataan benar atau tidak. Contohnya, jika ingin mengecek apakah nilai seorang siswa lebih dari atau sama dengan 75, hasil pengecekannya dapat berupa True atau False.

### 2.2.3 Program "Konstanta_2611533014"

```python
#Buat file dengan nama Konstanta_2611533014.py
#Program ini menggunakan konstanta untuk menghitung luas lingkaran
#nama variabel ditambah 4 digit nim terakhir contoh:jari_3014

from typing import Final
PI : Final = 3.14
print("pi:%f % (PI)")
jari_3014 = float(input('masukkan nilai jari-jari:'))
luas_3014 = PI * jari_3014
print ("Luas lingkaran dengan jari %.2f adalah %2f" %(jari_3014,luas_3014))
```

**Penjelasan:**
1. PI : Final = 3.14 → membuat konstanta PI dengan nilai tetap 3,14.
2. input() → digunakan untuk meminta pengguna memasukkan nilai jari-jari lingkaran.
3. float() → mengubah input menjadi angka desimal.
4. luas_3014 = PI * jari_3014 → menghitung luas lingkaran berdasarkan nilai PI dan jari-jari.
5. print() → menampilkan nilai dan hasil luas lingkaran.
6. Fungsi Final digunakan untuk menandakan bahwa nilai PI merupakan konstanta yang seharusnya tidak diubah. Program ini menunjukkan penggunaan konstanta, input, variabel, dan operasi matematika dalam Python.

### 2.2.4 Program "Input_2611533014"

```python
#buat file dengan nama Input_2611533014.py
#Buat program untuk menghitung luas persegi panjang
#Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1234
#Program ini menggunakan fungsi input()untuk meminta pengguna 
#memasukkan nilai panjang dan lebar persegi panjang.
#Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_3014 = input ('Masukkan nilai panjang:')
lebar_3014 = input("Masukkan nilai lebar: ")
luas_3014 = int(panjang_3014)* int(lebar_3014)
print("luas =",luas_3014)
```

### 2.2.5 Program "Boolean_2611533014"

```python
#Buat file dengan nama Boolean_2611533014.py
#Nama variabel ditambah 4 digit nim terakhir contoh:nilai_1234
#Deklarasi variabel dengan tipe data Boolean
is_lulus_3014 = True
is_cumlaude_3014 = True

# Menggunakan Boolean
nilai_3014 = 85
batas_lulus_3014 = 75

#Menentukan nilai boolean dari kondisi
status_kelulusan_3014 = nilai_3014 >= batas_lulus_3014 #hasilnya akan true

print("=== Check Kelulusan ===")
print("Nilai:",nilai_3014)
print("Apakah lulus?:",status_kelulusan_3014)
if is_lulus_3014 and is_cumlaude_3014:
        print("Selamat,Anda lulus dengan prediket Cum Laude")
```

**Penjelasan:**
1. is_lulus_3014 = True dan is_cumlaude_3014 = True → digunakan untuk menyimpan kondisi kelulusan dan predikat.
2. nilai_3014 = 85 dan batas_lulus_3014 = 75 → menyimpan nilai yang diperoleh dan nilai minimum untuk lulus.
3. status_kelulusan_3014 = nilai_3014 >= batas_lulus_3014 → membandingkan nilai dengan batas lulus. Karena 85 lebih besar dari 75, hasilnya adalah True.
4. print() → digunakan untuk menampilkan nilai dan status kelulusan.
5. if is_lulus_3014 and is_cumlaude_3014: → menggunakan operator and untuk mengecek apakah kedua kondisi bernilai True. Jika keduanya benar, program menampilkan ucapan bahwa pengguna lulus dengan predikat Cum Laude.
