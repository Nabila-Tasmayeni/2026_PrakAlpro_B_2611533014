## 2.1 Dasar Teori
Struktur percabangan adalah salah satu bagian dasar dalam pemrograman yang digunakan untuk membuat program mengambil keputusan berdasarkan suatu kondisi. Dengan percabangan, program dapat menjalankan perintah yang berbeda tergantung apakah suatu kondisi terpenuhi atau tidak.

Pada Python, struktur percabangan menggunakan kondisi yang menghasilkan nilai True atau False. Jika kondisi bernilai True, maka perintah yang berada di dalam percabangan akan dijalankan. Sebaliknya, jika kondisi bernilai False, program dapat menjalankan perintah lain atau melewati perintah tersebut.

1. **if**  
if digunakan untuk menjalankan perintah jika suatu kondisi terpenuhi. Jika kondisi tidak terpenuhi, perintah tersebut tidak dijalankan.

2. **if-else**  
if-else digunakan untuk menentukan dua kemungkinan. Jika kondisi terpenuhi, bagian if dijalankan, sedangkan jika tidak terpenuhi, bagian else dijalankan.

3. **if-elif-else**  
if-elif-else digunakan untuk memeriksa beberapa kondisi. Program akan menjalankan perintah berdasarkan kondisi yang terpenuhi.

4. **Multi-if**  
Multi-if menggunakan beberapa if untuk memeriksa kondisi secara terpisah. Setiap kondisi yang terpenuhi dapat dijalankan.

5. **Match-case**  
match-case digunakan untuk mencocokkan suatu nilai dengan beberapa pilihan yang telah ditentukan. Struktur ini cocok digunakan pada program yang memiliki beberapa pilihan.

### 2.2.4 Program "multi_if2_2611533014.py"

```python
#Buat nama file dengan nama multi_if2_2611533014.py
#Buat program untuk kondisional if
#Nama variabel ditambah 4 digit NIM terakhir contoh: ipk_1234
#Program ini menggunakan fungsi input ()
#Program Menghitung Diskon Belanja

# Input dan user
total_belanja_3014 = float(input("Input Total Belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 't')
input_member_3014 = input("Apakah Anda Member (y/t): ").strip().lower()
is_member_3014 = input_member_3014 in ["y", "t"]

# Input satus kode promo (mengecek apakah user mengetik 'y atau 't')
input_promo_3014 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3014 = input_promo_3014 in ["y", "t"]

total_diskon_persen_3014 = 0

# Multi-if terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3014 > 1000000:
    total_diskon_persen_3014 += 10  # Diskon total besar

if is_member_3014:
    total_diskon_persen_3014 += 5  # Diskon member

if kode_promo_valid_3014:
    total_diskon_persen_3014 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_3014 = total_belanja_3014 * (total_diskon_persen_3014 / 100)
total_bayar_3014 = total_belanja_3014 - nominal_diskon_3014

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_3014} (Rp {nominal_diskon_3014:.0f})")
print(f"Total Bayar   : Rp {total_bayar_3014:.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_3014}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan:**

1. Program meminta pengguna memasukkan total belanja. Nilai 80.000 disimpan dalam variabel total_belanja_3014.
2. Selanjutnya, program meminta pengguna memasukkan status member dan kode promo dengan pilihan 'y' atau 't'.
3. Program menggunakan beberapa if untuk mengecek setiap kondisi secara terpisah. Jadi, beberapa diskon bisa didapatkan sekaligus.
4. Jika total belanja lebih dari Rp1.000.000, pengguna mendapat diskon 10%. Karena total belanja hanya Rp80.000, diskon ini tidak didapatkan.
5. Pada bagian member, input 'y' membuat is_member_3014 bernilai True, sehingga mendapat diskon 5%.
6. Pada bagian kode promo, input 'y' membuat kode_promo_valid_3014 bernilai True, sehingga mendapat diskon 15%.
7. Jadi, dari input Rp80.000, y, y, total diskon yang didapatkan adalah 20% karena mendapat diskon member 5% dan promo 15%.
8. Setelah itu, program menghitung jumlah diskon dan mengurangi jumlah tersebut dari total belanja. Jadi, total yang harus dibayar adalah Rp64.000.

### 2.2.6 Program "nilai_2611533014.py"

```python
# Buat file dengan nama nilai_nim.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi nilai angka menjadi huruf

nilai_3014 = int(input("Input nilai angka: "))

if nilai_3014 >= 81:
    print("A")
elif nilai_3014 >= 70:
    print("B")
elif nilai_3014 >= 60:
    print("C")
elif nilai_3014 >= 50:
    print("D")
else:
    print("E")
```

4. Pada bagian if, program mengecek kondisi pertama, yaitu apakah nilai lebih dari atau sama dengan 81. Jika kondisi benar, program langsung menampilkan A.
5. Pada bagian elif, program mengecek kondisi berikutnya jika kondisi if sebelumnya tidak terpenuhi. Jadi, elif digunakan untuk memberikan pilihan kondisi lainnya.
6. Jika nilai 81 atau lebih, hasilnya A. Jika tidak, program lanjut mengecek elif berikutnya sampai menemukan kondisi yang sesuai.
7. Nilai 70–80 mendapatkan B, nilai 60–69 mendapatkan C, nilai 50–59 mendapatkan D, dan nilai di bawah 50 mendapatkan E.
8. Perbedaannya, if digunakan untuk mengecek kondisi pertama, sedangkan elif digunakan untuk mengecek kondisi berikutnya jika kondisi sebelumnya tidak terpenuhi.

### 2.2.7 Program "match_case1_2611533014.py"

```python
#Buat file dengan nama match_case1_nim.py
#Buat program untuk mach case
#Nama variabel ditambah 4 digit nim terakhir contoh:ipk_1234
#Program ini menggunakan fungsi input ()
#Program konversi angka menjadi nama bulan

bulan_3014 = int(input("Masukkan angka bulan(1-12): "))

match bulan_3014:
        case 1:
            print("Januari")
        case 2:
            print("Februari")
        case 3:
            print("Maret")
        case 4:
            print("April")
        case 5:
            print("Mei")
        case 6:
            print("Juni")
        case 7:
            print("Juli")
        case 8:
            print("Agustus")
        case 9:
            print("September")
        case 10:
            print("Oktober")
        case 11:
            print("November")
        case 12:
            print("Desember")
```

**Penjelasan:**

1. Program meminta pengguna memasukkan angka bulan dari 1 sampai 12 menggunakan fungsi input().
2. Angka yang dimasukkan disimpan dalam variabel bulan_3014 dan diubah menjadi bilangan bulat menggunakan int().
3. Program menggunakan match-case untuk mencocokkan angka yang dimasukkan dengan pilihan bulan yang tersedia.
4. Setiap case memiliki angka dan nama bulan yang berbeda. Misalnya, case 1 akan menampilkan Januari, case 2 menampilkan Februari, dan seterusnya sampai case 12 yang menampilkan Desember.
5. Jika angka yang dimasukkan sesuai dengan salah satu case, program akan menampilkan nama bulan yang sesuai.
6. Karena angka yang dimasukkan adalah 10, program mencocokkannya dengan case 10.
Setelah cocok, program menampilkan “Oktober” sebagai hasilnya.
7. Jika angka yang dimasukkan tidak sesuai dengan angka 1 sampai 12, program menjalankan case _ dan menampilkan pesan "Angka tidak valid".
8. Setelah nama bulan atau pesan kesalahan ditampilkan, program selesai.

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# Input data pengunjung
nama_3014 = input("Masukkan Nama Pengunjung : ")
umur_3014 = int(input("Input umur anda : "))
sim_3014 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].lower()
jumlah_tiket_3014 = int(input("Masukkan jumlah tiket : "))

# Validasi jumlah tiket
if jumlah_tiket_3014 <= 0:
    print("Peringatan: Jumlah tiket tidak valid!")

# Pilihan wahana
print("\nPilihan Paket Wahana (1-5):")
print("1. Safari Rimba         (Rp 50,000)")
print("2. Arung Jeram          (Rp 75,000)")
print("3. Motor ATV Ekstrim    (Rp 120,000)")
print("4. Roller Coaster Kilat (Rp 100,000)")
print("5. All-Access VIP       (Rp 220,000)")

paket_3014 = int(input("Masukkan nomor paket (1-5): "))

# Menentukan wahana dan harga
match paket_3014:
    case 1:
        nama_wahana_3014 = "Wahana Safari Rimba"
        harga_satuan_3014 = 50000
    case 2:
        nama_wahana_3014 = "Wahana Arung Jeram"
        harga_satuan_3014 = 75000
    case 3:
        nama_wahana_3014 = "Wahana Motor ATV Ekstrim"
        harga_satuan_3014 = 120000
    case 4:
        nama_wahana_3014 = "Wahana Roller Coaster Kilat"
        harga_satuan_3014 = 100000
    case 5:
        nama_wahana_3014 = "Wahana All-Access VIP"
        harga_satuan_3014 = 220000
    case _:
        print("Paket wahana tidak valid!")
        raise SystemExit

# Input member dan promo
is_member_3014 = input("Apakah Anda member? (y/t): ").strip().lower()
kode_promo_valid_3014 = input("Apakah kode promo valid? (y/t): ").strip().lower()

# Validasi izin wahana
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_3014 == 3:
    if umur_3014 >= 17 and sim_3014 == "y":
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
    elif umur_3014 >= 17 and sim_3014 != "y":
        print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
    elif umur_3014 < 17 and sim_3014 == "y":
        print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
    else:
        print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
else:
    if umur_3014 >= 10:
        print("Status Akses: Anda memenuhi batas umur wahana.")
    else:
        print("Status Akses: Anda belum cukup umur untuk wahana ini.")

# Menghitung subtotal
subtotal_3014 = harga_satuan_3014 * jumlah_tiket_3014

# Menghitung diskon
total_diskon_persen_3014 = 0

if subtotal_3014 >= 200000:
    total_diskon_persen_3014 += 10

if is_member_3014 in ["y", "ya"]:
    total_diskon_persen_3014 += 5

if kode_promo_valid_3014 in ["y", "ya"]:
    total_diskon_persen_3014 += 15

if jumlah_tiket_3014 >= 5:
    total_diskon_persen_3014 += 5

# Menghitung pembayaran
nominal_diskon_3014 = subtotal_3014 * (total_diskon_persen_3014 / 100)
total_bayar_3014 = subtotal_3014 - nominal_diskon_3014

# Menampilkan rincian pembayaran
print("\n--- RINCIAN PEMBAYARAN ---")
print("Nama Pengunjung : ", nama_3014)
print("Wahana          : ", nama_wahana_3014)
print(f"Subtotal Belanja : Rp {subtotal_3014:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_3014}% (Rp {nominal_diskon_3014:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_3014:,.0f}")

# Evaluasi total bayar
if total_bayar_3014 > 300000:
    print("Catatan Layanan  : Selamat! Anda berhak mendapatkan Souvenir Gratis.")
else:
    print("Catatan Layanan  : Terima kasih telah berkunjung.")

print("Program Selesai")
```

**Penjelasan:**
1. Input data pengunjung  
Program meminta pengguna memasukkan nama, umur, kepemilikan SIM C, dan jumlah tiket.
2. Percabangan if  
if digunakan untuk mengecek jumlah tiket. Jika jumlah tiket 0 atau kurang, program akan menampilkan peringatan bahwa jumlah tiket tidak valid.
3. Percabangan match-case  
match-case digunakan untuk menentukan wahana berdasarkan nomor yang dipilih. case 1 sampai case 5 digunakan untuk menentukan nama dan harga wahana. case _ digunakan jika nomor yang dimasukkan tidak sesuai dengan pilihan.
4. Percabangan if-elif-else  
Pada wahana ATV, if-elif-else digunakan untuk mengecek umur dan kepemilikan SIM C. Operator and digunakan untuk menggabungkan kondisi, sedangkan != digunakan untuk mengecek kondisi yang tidak sama dengan y.
5. Percabangan if-else  
Untuk wahana selain ATV, if-else digunakan untuk mengecek apakah umur pengunjung sudah mencapai 10 tahun. Jika memenuhi, pengunjung dinyatakan boleh mengikuti wahana, dan jika tidak maka dinyatakan belum cukup umur.
6. Perhitungan subtotal  
Program menghitung subtotal dengan mengalikan harga satuan wahana dengan jumlah tiket yang dibeli.
7. Multi-if untuk diskon  
Beberapa if digunakan secara terpisah untuk mengecek syarat diskon. Diskon diberikan berdasarkan total belanja, status member, kode promo, dan jumlah tiket. Jika beberapa syarat terpenuhi, diskonnya dapat ditambahkan.
8. Menghitung total pembayaran  
Program menghitung jumlah diskon dalam rupiah, kemudian mengurangi diskon tersebut dari subtotal untuk mendapatkan total yang harus dibayar.
9. If-else untuk souvenir  
if-else digunakan untuk mengecek total pembayaran. Jika total bayar lebih dari Rp300.000, pengunjung mendapat informasi souvenir gratis. Jika tidak, program menampilkan ucapan terima kasih.
