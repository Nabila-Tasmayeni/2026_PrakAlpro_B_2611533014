## 2.1 Dasar Teori
Praktikum mengenai Pengenalan Python, VS Code, dan Koneksi ke GitHub ini mempelajari dasar-dasar cara menulis kode dengan Python, menggunakan aplikasi VS Code sebagai tempat mengetik kode, serta memanfaatkan GitHub sebagai wadah untuk menyimpan dan mengorganisir file codingan kita secara online.

### Python
Python adalah salah satu bahasa pemrograman tingkat tinggi yang sangat populer karena kodenya ringkas dan tidak rumit, sehingga ramah untuk pemula yang baru belajar coding. Python punya banyak fungsi di dunia teknologi, mulai dari membuat aplikasi, mengolah data, hingga membangun sistem perangkat lunak yang kompleks.

### Visual Studio Code
Visual Studio Code (VS Code) adalah aplikasi penyunting kode (*code editor*) yang dipakai untuk mengetik, memperbaiki, dan menguji jalannya sebuah program. Aplikasi ini bisa membaca banyak bahasa pemrograman termasuk Python, serta dilengkapi fasilitas bawaan yang sangat membantu seperti terminal langsung dan sistem pelacak eror (*debugging*).

### Git
Git merupakan sebuah sistem pelacak riwayat (*version control*) yang bertugas mencatat setiap perubahan yang kita lakukan pada file codingan. Melalui Git, seorang programmer bisa melihat kembali versi lama dari kodenya dan tahu bagian mana saja yang pernah diubah sebelumnya.

### GitHub
GitHub adalah sebuah website atau platform online yang digunakan sebagai tempat menampung data-data dari Git. Dengan GitHub, kita bisa mencadangkan (*backup*) kode kita di internet, membuka file tugas dari komputer mana saja, serta bekerja kelompok dengan sesama programmer tanpa takut file tertimpa.

### Repository
Repository (sering disebut Repo) adalah folder penyimpanan khusus yang berisi seluruh file kode program beserta dokumen pendukung dari suatu project. Folder penyimpanan ini bisa berada di dalam komputer kita sendiri (lokal) atau diunggah ke internet melalui GitHub (online).

### Commit dan Push
Commit adalah perintah atau cara untuk mengunci dan menyimpan riwayat perubahan kode yang baru saja kita ketik di dalam komputer lokal. Sedangkan Push adalah proses mengirimkan file-file yang sudah di-commit tadi dari komputer kita ke dalam akun GitHub agar tersimpan secara online.
